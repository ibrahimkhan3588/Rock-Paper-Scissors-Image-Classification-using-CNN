```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import os

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D
from tensorflow.keras.layers import Flatten, Dense, Dropout
from tensorflow.keras.preprocessing.image import ImageDataGenerator

print("All libraries imported successfully!")
print("TensorFlow Version:", tf.__version__)
```

```python
import os

base_path = r"C:\Users\ibrahim khan\Desktop\DL_RPS_Project\dataset\Rock-Paper-Scissors_Dataset"

train_path = os.path.join(base_path, "train")
validation_path = os.path.join(base_path, "validation")
test_path = os.path.join(base_path, "test")

print("Training path:", train_path)
print("Validation path:", validation_path)
print("Testing path:", test_path)
```

```python
print("Main dataset folders:")

print(os.listdir(base_path))

print("\nTraining folders:")
print(os.listdir(train_path))

print("\nValidation folders:")
print(os.listdir(validation_path))

print("\nTesting folders:")
print(os.listdir(test_path))
```

```python
classes = ['paper', 'rock', 'scissors']

for class_name in classes:
    train_count = len(os.listdir(os.path.join(train_path, class_name)))
    validation_count = len(os.listdir(os.path.join(validation_path, class_name)))
    test_count = len(os.listdir(os.path.join(test_path, class_name)))

    print(f"\n{class_name.upper()}")
    print("Training:", train_count)
    print("Validation:", validation_count)
    print("Testing:", test_count)
```

```python
import matplotlib.pyplot as plt
from PIL import Image

plt.figure(figsize=(12, 4))

for i, class_name in enumerate(classes):
    image_folder = os.path.join(train_path, class_name)
    image_name = os.listdir(image_folder)[0]
    image_path = os.path.join(image_folder, image_name)

    img = Image.open(image_path)

    plt.subplot(1, 3, i + 1)
    plt.imshow(img)
    plt.title(class_name.upper())
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
for class_name in classes:

    image_folder = os.path.join(train_path, class_name)
    image_name = os.listdir(image_folder)[0]
    image_path = os.path.join(image_folder, image_name)

    img = Image.open(image_path)

    print(class_name.upper(), "Image Size:", img.size)
```

```python
from PIL import Image

corrupted_images = []

for class_name in classes:

    image_folder = os.path.join(train_path, class_name)

    for image_name in os.listdir(image_folder):

        image_path = os.path.join(image_folder, image_name)

        try:
            img = Image.open(image_path)
            img.verify()

        except Exception:
            corrupted_images.append(image_path)

print("Total corrupted images:", len(corrupted_images))
```

```python
IMG_SIZE = 150
BATCH_SIZE = 32

print("Image Size:", IMG_SIZE, "x", IMG_SIZE)
print("Batch Size:", BATCH_SIZE)
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    zoom_range=0.2,
    horizontal_flip=True,
    width_shift_range=0.1,
    height_shift_range=0.1
)

print("Training data generator created successfully!")
```

```python
train_data = train_datagen.flow_from_directory(
    train_path,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=True
)
```

```python
validation_datagen = ImageDataGenerator(
    rescale=1./255
)

validation_data = validation_datagen.flow_from_directory(
    validation_path,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False
)
```

```python
test_datagen = ImageDataGenerator(
    rescale=1./255
)

test_data = test_datagen.flow_from_directory(
    test_path,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False
)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout

model = Sequential()

# First convolutional layer
model.add(Conv2D(32, (3, 3), activation='relu',
                 input_shape=(IMG_SIZE, IMG_SIZE, 3)))
model.add(MaxPooling2D(pool_size=(2, 2)))

# Second convolutional layer
model.add(Conv2D(64, (3, 3), activation='relu'))
model.add(MaxPooling2D(pool_size=(2, 2)))

# Flatten layer
model.add(Flatten())

# Fully connected layers
model.add(Dense(128, activation='relu'))
model.add(Dropout(0.5))

# Output layer: Paper, Rock, Scissors
model.add(Dense(3, activation='softmax'))

print("CNN model created successfully!")
```

```python
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

print("CNN model compiled successfully!")
```

```python
history = model.fit(
    train_data,
    epochs=10,
    validation_data=validation_data
)
```

```python
test_loss, test_accuracy = model.evaluate(test_data)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy * 100, "%")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')

plt.title('CNN Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')

plt.title('CNN Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np

predictions = model.predict(test_data)

predicted_classes = np.argmax(predictions, axis=1)
actual_classes = test_data.classes

print("Total predictions:", len(predicted_classes))
print("Total actual labels:", len(actual_classes))
print("Predictions generated successfully!")
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(actual_classes, predicted_classes)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=list(test_data.class_indices.keys())
)

disp.plot(cmap='Blues')
plt.title("CNN Confusion Matrix")
plt.show()
```

```python
from sklearn.metrics import classification_report

report = classification_report(
    actual_classes,
    predicted_classes,
    target_names=list(test_data.class_indices.keys())
)

print(report)
```

```python
model.save("rock_paper_scissors_cnn.keras")

print("Model saved successfully!")
```
